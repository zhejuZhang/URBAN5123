```python
#import some necessary librairies
import numpy as np # linear algebra
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)
import matplotlib.pyplot as plt  # Matlab-style plotting
import seaborn as sns
from scipy.stats import norm
from sklearn.preprocessing import StandardScaler
from scipy import stats
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.linear_model import Ridge, Lasso, ElasticNet
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
# import the pandas dataframe 
train= pd.read_csv('data/train.csv')
test = pd.read_csv('data/test.csv')
train.head()
```

# Making sense of the data

```python
## Question 1: How many features are in the training and testing datasets?
train_features = train.shape[1] - 1  # exclude the target column price_doc
test_features = test.shape[1]

print(f"Training features: {train_features}")
print(f"Testing features:  {test_features}")
```

```python
##Question 2: How to carry out normality test on the target variable?

target = train["price_doc"].dropna()

# Histogram and quantile-quantile plot
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
sns.histplot(target, bins=50, kde=True)
plt.title("price_doc distribution")

plt.subplot(1, 2, 2)
stats.probplot(target, dist="norm", plot=plt)
plt.title("Q-Q plot")
plt.tight_layout()
plt.show()

# D'Agostino-Pearson test: reject H0 (normal distribution) when p < 0.05
k2_statistic, p_value = stats.normaltest(target)
print(f"D'Agostino-Pearson statistic = {k2_statistic:.4f}")
print(f"p-value = {p_value:.3e}")
```

```python
## Question 3: How to convert the target variable into a more normal distribution?
train["price_doc_log"] = np.log1p(train["price_doc"])

plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
sns.histplot(train["price_doc_log"], bins=50, kde=True)
plt.title("log1p(price_doc)")

plt.subplot(1, 2, 2)
stats.probplot(train["price_doc_log"], dist="norm", plot=plt)
plt.title("Q-Q plot after log1p")
plt.tight_layout()
plt.show()

k2_statistic, p_value = stats.normaltest(train["price_doc_log"])
print(f"D'Agostino-Pearson statistic after log1p = {k2_statistic:.4f}")
print(f"p-value = {p_value:.3e}")
```

# Correlation Analysis

```python
## Question 4: How to plot the pairwise correlation matrix?
numeric_train = train.select_dtypes(include=np.number)
corrmat = numeric_train.corr()

plt.figure(figsize=(12, 9))
sns.heatmap(corrmat, cmap="RdBu_r", center=0, vmin=-1, vmax=1,
            square=True)
plt.title("Pairwise correlation matrix")
plt.tight_layout()
plt.show()
```

```python
## Question 5: Plot the pairwise correlation matrix for the top 10 numeric features.
# Rank features by absolute Pearson correlation with log-transformed price_doc.
numeric_for_corr = numeric_train.drop(columns=["price_doc_log"], errors="ignore")
target_corr = numeric_for_corr.corrwith(train["price_doc_log"]).dropna()
top10 = target_corr.abs().nlargest(10).index

print("Top 10 features by absolute Pearson correlation:")
print(pd.DataFrame({"correlation": target_corr[top10]}))

top10_with_target = top10.tolist() + ["price_doc_log"]
cm = train[top10_with_target].corr()
plt.figure(figsize=(12, 10))
sns.heatmap(cm, cbar=True, annot=True, square=True, fmt=".2f",
            cmap="RdBu_r", center=0, vmin=-1, vmax=1)
plt.title("Top 10 correlations with log-transformed price_doc")
plt.tight_layout()
plt.show()

# Question 5.1: Absolute correlation captures both strong positive and strong
# negative linear relationships. It also provides a consistent numeric ranking.
```

```python
## Question 6: How to plot the pairwise scatterplot?
pairplot_columns = top10[:5].tolist() + ["price_doc_log"]
sns.pairplot(train[pairplot_columns].dropna(), corner=True,
             plot_kws={"alpha": 0.25, "s": 12})
plt.suptitle("Pairwise scatterplots for the strongest numeric features", y=1.02)
plt.show()
```

# Feature Engineering

```python
## Question 7: How to concatenate the train and test datasets?
train_clean = train.drop(columns=["price_doc_log"]).copy()
test_clean = test.copy()
train_clean["__is_train"] = 1
test_clean["__is_train"] = 0

combined = pd.concat([train_clean.drop(columns=["price_doc"]), test_clean],
                     ignore_index=True, sort=False)
print("Concatenated shape:", combined.shape)
print("Training rows:", train_clean.shape[0], "Testing rows:", test_clean.shape[0])
```

```python
## Question 8: Drop features with a high missing ratio and impute remaining values.
missing_ratio = combined.drop(columns=["__is_train"], errors="ignore").isna().mean()
high_missing = missing_ratio[missing_ratio > 0.40].index.tolist()

print(f"Dropping {len(high_missing)} features with >40% missing values:")
print(high_missing)
combined = combined.drop(columns=high_missing)

# Median for numeric features and the most frequent value for categorical features.
numeric_columns = combined.select_dtypes(include=[np.number]).columns
numeric_columns = numeric_columns.drop("__is_train", errors="ignore")
categorical_columns = combined.select_dtypes(include=["object", "string", "category"]).columns

for column in numeric_columns:
    combined[column] = combined[column].fillna(combined[column].median())

for column in categorical_columns:
    combined[column] = combined[column].fillna("Missing")

print("Remaining missing values:", int(combined.isna().sum().sum()))
```

```python
## Question 9: Detect and remove outliers.
# The target is skewed, so use IQR filtering to remove extreme transaction prices.
q1 = train_clean["price_doc"].quantile(0.25)
q3 = train_clean["price_doc"].quantile(0.75)
iqr = q3 - q1
target_lower = q1 - 1.5 * iqr
target_upper = q3 + 1.5 * iqr
target_mask = train_clean["price_doc"].between(target_lower, target_upper)

# Also remove clearly invalid or extremely large total-area observations.
if "full_sq" in train_clean.columns:
    area_mask = train_clean["full_sq"].between(1, 500)
else:
    area_mask = pd.Series(True, index=train_clean.index)

outlier_mask = target_mask & area_mask
removed_rows = int((~outlier_mask).sum())
train_clean = train_clean.loc[outlier_mask].copy()
print(f"Removed {removed_rows} outlier rows")
print(f"Remaining training rows: {len(train_clean)}")

# Rebuild the combined dataset because the training row count changed.
train_clean["__is_train"] = 1
test_clean["__is_train"] = 0
combined = pd.concat([
    train_clean.drop(columns=["price_doc"]),
    test_clean
], ignore_index=True, sort=False)

combined = combined.drop(columns=[column for column in high_missing if column in combined.columns],
                         errors="ignore")
numeric_columns = combined.select_dtypes(include=[np.number]).columns.drop("__is_train", errors="ignore")
categorical_columns = combined.select_dtypes(include=["object", "string", "category"]).columns
for column in numeric_columns:
    combined[column] = combined[column].fillna(combined[column].median())
for column in categorical_columns:
    combined[column] = combined[column].fillna("Missing")
```

```python
## Question 10: Convert categorical features into numeric features.
if "timestamp" in combined.columns:
    timestamp = pd.to_datetime(combined["timestamp"], errors="coerce")
    combined["sale_year"] = timestamp.dt.year
    combined["sale_month"] = timestamp.dt.month
    combined["sale_day"] = timestamp.dt.day
    combined = combined.drop(columns=["timestamp"])

combined = pd.get_dummies(combined, drop_first=True)
print("Shape after categorical encoding:", combined.shape)
```

# Training 

```python
## Question 11: Perform a training-validation split after feature engineering.
train_mask = combined["__is_train"].astype(bool)
X = combined.drop(columns=["__is_train"])
X_train_full = X.loc[train_mask].copy()
X_test_full = X.loc[~train_mask].copy()
y_full = np.log1p(train_clean["price_doc"])

X_train, X_valid, y_train, y_valid = train_test_split(
    X_train_full, y_full, test_size=0.20, random_state=42
)
print("Training shape:", X_train.shape)
print("Validation shape:", X_valid.shape)
print("Submission feature shape:", X_test_full.shape)
```

```python
## Question 12: Fine-tune the parameters of each model.
tuning_size = min(8000, len(X_train))
X_tune = X_train.iloc[:tuning_size]
y_tune = y_train.iloc[:tuning_size]

model_candidates = {
    "ridge": (
        Pipeline([("scale", StandardScaler()), ("model", Ridge())]),
        {"model__alpha": [0.1, 1.0, 10.0, 100.0]}
    ),
    "lasso": (
        Pipeline([("scale", StandardScaler()),
                  ("model", Lasso(max_iter=10000, tol=1e-4, random_state=42))]),
        {"model__alpha": [0.0001, 0.001, 0.01]}
    ),
    "elastic_net": (
        Pipeline([("scale", StandardScaler()),
                  ("model", ElasticNet(max_iter=10000, tol=1e-4, random_state=42))]),
        {"model__alpha": [0.0001, 0.001, 0.01],
         "model__l1_ratio": [0.25, 0.75]}
    )
}

tuned_models = {}
for name, (pipeline, params) in model_candidates.items():
    search = GridSearchCV(pipeline, params, cv=3,
                          scoring="neg_root_mean_squared_error",
                          n_jobs=-1)
    search.fit(X_tune, y_tune)
    tuned_models[name] = search.best_estimator_
    print(name, "best parameters:", search.best_params_,
          "CV RMSE:", -search.best_score_)
```

```python
## Question 13: Difference between Ridge and Lasso; when to use each.
# Ridge uses an L2 penalty: it shrinks coefficients but usually keeps all features.
# Lasso uses an L1 penalty: it can shrink coefficients exactly to zero, effectively
# performing feature selection.
# Use Ridge when many correlated features are expected and a stable dense model is useful.
# Use Lasso when a sparse model or automatic feature selection is desirable.
# Elastic Net combines L1 and L2 and is a useful compromise for correlated features.

for name in ["ridge", "lasso", "elastic_net"]:
    model = tuned_models[name]
    linear_model = model.named_steps["model"]
    print(name, "non-zero coefficients:", int(np.count_nonzero(linear_model.coef_)))
```

```python
## Question 14: Ensemble the results of several models.
validation_predictions = {}
submission_predictions = {}

for name, model in tuned_models.items():
    model.fit(X_train, y_train)
    validation_predictions[name] = model.predict(X_valid)
    submission_predictions[name] = model.predict(X_test_full)

# Simple arithmetic mean gives each tuned model equal weight.
ensemble_valid_log = np.mean(list(validation_predictions.values()), axis=0)
ensemble_test_log = np.mean(list(submission_predictions.values()), axis=0)

print("Validation RMSE (log scale):",
      np.sqrt(mean_squared_error(y_valid, ensemble_valid_log)))
```

# Testing 

```python
# Question 15: Evaluate and submit predictions for the test dataset.
valid_original = np.expm1(y_valid)

def evaluation_metrics(y_true, y_pred_original):
    y_pred_original = np.clip(y_pred_original, 0, None)
    rmsle = np.sqrt(mean_squared_error(np.log1p(y_true), np.log1p(y_pred_original)))
    return {
        "RMSE": np.sqrt(mean_squared_error(y_true, y_pred_original)),
        "MAE": mean_absolute_error(y_true, y_pred_original),
        "R2": r2_score(y_true, y_pred_original),
        "RMSLE": rmsle
    }

print("Validation metrics:")
for name, log_predictions in validation_predictions.items():
    print(name, evaluation_metrics(valid_original, np.expm1(log_predictions)))

ensemble_valid_original = np.expm1(ensemble_valid_log)
print("ensemble", evaluation_metrics(valid_original, ensemble_valid_original))

ensemble_test_original = np.expm1(ensemble_test_log)
ensemble_test_original = np.clip(ensemble_test_original, 0, None)
submission = pd.DataFrame({
    "id": test["id"].astype(int),
    "price_doc": ensemble_test_original
})
submission.to_csv("submission.csv", index=False)
print("Saved submission.csv")
submission.head()
```
